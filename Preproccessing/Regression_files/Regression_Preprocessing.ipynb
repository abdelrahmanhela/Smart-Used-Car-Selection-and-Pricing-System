```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/abdelrahmanhela/Smart-Used-Car-Selection-and-Pricing-System/refs/heads/main/Data%20Cleaning/cars_clean.csv")
df.head()
```

```python
df.info()
```

## Separate Features & Target

```python
X = df.drop(columns=['price_egp', 'id'])
y = df['price_egp']
```

```python
# Hyundai: 1200 سيارة.

# Kia: 900 سيارة.

# Nissan: 600 سيارة.

# Other: 300 سيارة.
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=45 ,stratify=X['make']

)
```

```python
color_mode = X_train['color'].mode()[0]

X_train['color'] = X_train['color'].fillna(color_mode)
X_test['color'] = X_test['color'].fillna(color_mode)
```

```python
X_train['km'] = X_train['km'].fillna(X_train['km'].median())
X_test['km'] = X_test['km'].fillna(X_train['km'].median())
print(X_train['km'].isna().sum())
print(X_test['km'].isna().sum())
```

## Encoding

**OHE**

```python
ohe_cols = ['make', 'transmission', 'seller_type', 'fuel_type', 'color']
freq_cols = ['model', 'city']
```

```python
ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

X_train_ohe = ohe.fit_transform(X_train[ohe_cols])
X_test_ohe = ohe.transform(X_test[ohe_cols])

ohe_names = ohe.get_feature_names_out(ohe_cols)

X_train_ohe = pd.DataFrame(
    X_train_ohe,
    columns=ohe_names,
    index=X_train.index
)

X_test_ohe = pd.DataFrame(
    X_test_ohe,
    columns=ohe_names,
    index=X_test.index
)
```

```python
X_train = X_train.drop(columns=ohe_cols)
X_test = X_test.drop(columns=ohe_cols)

X_train = pd.concat([X_train, X_train_ohe], axis=1)
X_test = pd.concat([X_test, X_test_ohe], axis=1)
```

```python
X_train
```

**Frequency**

```python
freq_cols = ['model', 'city']
for col in freq_cols:
    freq = X_train[col].value_counts(normalize=True)

    X_train[col] = X_train[col].map(freq)
    X_test[col] = X_test[col].map(freq).fillna(0)
```

```python
X_train
```

```python
print(X_train.head())
print(X_train.dtypes)
```

## Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scale_col=['km','year']
X_train_scaled = scaler.fit_transform(X_train[scale_col])
X_test_scaled = scaler.transform(X_test[scale_col])

X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=scale_col,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=scale_col,
    index=X_test.index
)
```

```python
X_train['km']=X_train_scaled['km']
X_test['km']=X_test_scaled['km']
X_train['year']=X_train_scaled['year']
X_test['year']=X_test_scaled['year']
```
