```python
import pandas as pd 
```

```python
CarsDirty = pd.read_csv("../Egypt_Used_Cars_clean.csv")
CarsDirty.info()
CarsDirty['exterior_color'].unique()
```

```python
CarsClean = CarsDirty.copy()
# Standarizing The Null Since The Data Referes to it as Unknown 
for col in CarsDirty.columns:
    CarsClean[col] = CarsDirty[col].replace("Unknown",pd.NA)
CarsClean.info()
```

```python
CarsClean.describe(include='all')
```

```python
CarsClean = CarsClean.drop(columns=['listing_id','url'])
```

```python
# Contains alot of null and Irrelevant to The Price Target 
CarsClean.drop(columns=['exterior_color'],inplace=True) 
```

```python
# Removing Duplicates To Avoid Overfitting
CarsClean.drop_duplicates(inplace=True)
```

```python
# Removing All Non-Used Cars Since We are only interested in Used Cars
CarsClean = CarsClean[CarsClean['condition'] == "Used"]
```

```python
# Detecting Outliers in Kilometers and Age Columns
CarsClean[((CarsClean['kilometers'] > 100000) & (CarsClean['age'] <= 1))].info()
```

```python
CarsClean.to_csv("CleanData.csv", index=False)
```
